# Ejercicio 
Una cepa de E. Coli tiene una tasa de crecimiento tal que un individuo se duplica cada 30 minutos.
Calcula en cuanto tiempo esta cepa: 
a) Llenaría el volúmen de la Tierra (A partir del radio de la Tierra y cosiderandola como una esfera hueca)
b) Cubriría la superficie de la Tierra
c) Repite el calculo para una tasa de crecimeinto 
    c.1) Se duplica cada hora 
    c.2) Se duplica cada media hora 
    c.3) Se duplicada cada 1/4 de hora 

```python
import math as mt
```

```python
# a) Llenar el volumen de la Tierra - Saber la cantidad de bacterias que puede almacenar la Tierra
N = (6371e3)**3 / (0.5e-6)**3
print(N)
```

```python
# Generaciones de bacterias necesarias 
T = mt.log(N , 2) 
print(T)
```

```python
# Tiempo (se duplica cada 30 minutos) 
B = T * 30
Bh = B / 60
print(Bh)
```

```python
# b) Cubrir la superficie de la Tierra 
Ns = (6371e3)**2 / (0.5e-6)**2
print(Ns)
```

```python
# Generarciones de bacterias necesarias
Ts = mt.log(Ns, 2)
print(Ts)
```

```python
#Tiempo (se duplica cada 30 minutos)
Bs = Ts * 30
Bhs = Bs / 60
print(Bhs)
```

```python
# c) Calculo según las tasas de crecimiento (usando el volumen)
# Minutos si de duplica cada hora 
Tc1 = T * 60
print(Tc1)
# Minutos si se duplicada cada media hora 
Tc2 = T * 30
print(Tc2)
# Minutos si se duplica cada 1/4 de hora
Tc3 = T * 15
print(Tc3)
```

```python
# Calculo según las tasas de crecimiento (usando la superficie)
# Minutos si de duplica cada hora 
Tsc1 = Ts * 60
print(Tsc1)
# Minutos si se duplicada cada media hora 
Tsc2 = Ts * 30
print(Tsc2)
# Minutos si se duplica cada 1/4 de hora
Tsc3 = Ts * 15
print(Tsc3)
```
